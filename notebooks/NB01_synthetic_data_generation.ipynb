```python
import pandas as pd
import numpy as np
import random
import string
import os

RAW_DIR = '/content/drive/MyDrive/Portfolio/LuminaLearningGroup/data/raw'
os.makedirs(RAW_DIR, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

VERTICALS = ["Pharma", "Telecom", "Banking/Finance", "ERP/Software", "Call Center BPO"]

# C2.1. role_id -> (role_name, career_grade, total_count, per_vertical_count)
ROLE_SPEC = {
    "ROLE01": {"name": "Instructional Design Lead",   "grade": "CG60", "total": 50,  "per_vertical": 10},
    "ROLE02": {"name": "Instructional Design Manager", "grade": "CG70", "total": 25,  "per_vertical": 5},
    "ROLE03": {"name": "Content Designer",             "grade": "CG50", "total": 350, "per_vertical": 70},
    "ROLE04": {"name": "Solution Architect",           "grade": "CG80", "total": 5,   "per_vertical": 1},
}

# C2.2. Office assignment rule:
#  - ROLE01/02/03 (425 total) -> all Pune (OFF001)
#  - ROLE04 (5 total) -> 1 Pune, 2 Austin, 2 Manchester
ROLE04_OFFICE_SEQUENCE = ["OFF001", "OFF002", "OFF002", "OFF003", "OFF003"]

# Tenure distribution by career grade — triangular(low, mode, high), capped at 15
# (LLG founded 2011; project "now" = 2026 -> 15-year ceiling)
TENURE_PARAMS = {
    "CG50": (0, 2, 8),
    "CG60": (1, 4, 12),
    "CG70": (3, 7, 15),
    "CG80": (5, 10, 15),
}

def sample_tenure(grade):
    low, mode, high = TENURE_PARAMS[grade]
    val = np.random.triangular(low, mode, high)
    return int(min(15, max(0, round(val))))

def sample_certification_count():
    # Uniform baseline (vertical-domain cert + tool certs, e.g. Storyline/HTML) — everyone has this,
    # acquired at whatever point in their career they passed through CG50-70 (Solution Architects
    # carry it forward from their CG70 days per user confirmation).
    baseline = random.choice([3, 4])
    # Variable: DS/ML/NLP/GenAI upskilling certs, reflecting progress on K2.1
    variable = random.randint(0, 5)
    return baseline + variable

# ---------------------------------------------------------------
# C2.3. Build employees
# ---------------------------------------------------------------
employee_rows = []
emp_counter = 1

for role_id, spec in ROLE_SPEC.items():
    role04_idx = 0
    for vertical in VERTICALS:
        for _ in range(spec["per_vertical"]):
            employee_id = f"EMP{emp_counter:04d}"

            if role_id == "ROLE04":
                office_id = ROLE04_OFFICE_SEQUENCE[role04_idx]
                role04_idx += 1
            else:
                office_id = "OFF001"

            employee_rows.append({
                "employee_id": employee_id,
                "office_id": office_id,
                "role_id": role_id,
                "vertical": vertical,
                "tenure_years": sample_tenure(spec["grade"]),
                "certification_count": sample_certification_count(),
            })
            emp_counter += 1

employees = pd.DataFrame(employee_rows)
# ---------------------------------------------------------------
# 09 Oct 2026
# C2.3a. Re-key employee_id by joining order (realism fix, NB03 thread)
# Why: IDs were assigned role by role, so they ignored tenure. Real HR systems
# assign IDs at hire, so low IDs = longest tenure, and gaps belong to
# other departments and leavers.
# ---------------------------------------------------------------
EMP_ID_POOL = 1400   # ~1,000 current org headcount + leavers since 2011

employees["hire_year"] = 2026 - employees["tenure_years"]
employees = (employees.sample(frac=1, random_state=SEED)        # random order within the same hire year
                      .sort_values("hire_year", kind="stable")
                      .reset_index(drop=True))
id_numbers = sorted(random.sample(range(1, EMP_ID_POOL + 1), len(employees)))
employees["employee_id"] = [f"EMP{n:04d}" for n in id_numbers]
employees = employees.drop(columns="hire_year")

# ---------------------------------------------------------------
# C2.4. Build employees_pii
# ---------------------------------------------------------------
FIRST_NAMES = [
    "Aarav", "Priya", "Rohan", "Ananya", "Vikram", "Meera", "Arjun", "Kavya",
    "James", "Emily", "Michael", "Sarah", "David", "Jessica", "Daniel", "Laura",
    "Oliver", "Sophie", "Liam", "Charlotte", "Noah", "Amelia", "Ethan", "Grace",
    "Fatima", "Omar", "Layla", "Yusuf", "Chen", "Wei", "Mei", "Jun",
    "Carlos", "Sofia", "Diego", "Valentina", "Kwame", "Amara", "Kofi", "Zola",
]

def random_last_name():
    return "".join(random.choices(string.ascii_uppercase, k=3))

pii_rows = []
for emp_id in employees["employee_id"]:
    first = random.choice(FIRST_NAMES)
    last = random_last_name()
    full_name = f"{first} {last}"
    email = f"{first.lower()}.{last.lower()}@lumina-learning.com"
    pii_rows.append({
        "employee_id": emp_id,
        "first_name": first,
        "last_name": last,
        "full_name": full_name,
        "email": email,
    })

employees_pii = pd.DataFrame(pii_rows)

# ---------------------------------------------------------------
# C2.5. Save
# ---------------------------------------------------------------
employees.to_csv(os.path.join(RAW_DIR, "employees.csv"), index=False)
employees_pii.to_csv(os.path.join(RAW_DIR, "employees_pii.csv"), index=False)

# ---------------------------------------------------------------
# C2.6. Sanity checks
# ---------------------------------------------------------------
print(f"Total employees: {len(employees)}")
print(f"Total employees_pii: {len(employees_pii)}")
print()
print("Counts by role_id:")
print(employees["role_id"].value_counts())
print()
print("Counts by office_id:")
print(employees["office_id"].value_counts())
print()
print("Counts by vertical:")
print(employees["vertical"].value_counts())
print()
print("Office breakdown for ROLE04 (Solution Architect):")
print(employees[employees["role_id"] == "ROLE04"][["employee_id", "office_id", "vertical"]])
print()
print("Tenure stats by role_id:")
print(employees.groupby("role_id")["tenure_years"].agg(["min", "mean", "max"]))
print()
print("Certification count stats by role_id:")
print(employees.groupby("role_id")["certification_count"].agg(["min", "mean", "max"]))
print()
print("C2.3a check: IDs follow joining order (first = longest tenure, last = newest):")
print(employees[["employee_id", "role_id", "tenure_years"]].head(3).to_string(index=False))
print(employees[["employee_id", "role_id", "tenure_years"]].tail(3).to_string(index=False))
```

### __Cluster 3__

- **Build assessment_questions (16 rows):** 10 proficiency questions (one per competency) and 6 COM-B questions.

- **COM-B questions (6):**
- COM-B (Capability, Opportunity, Motivation) is a behavioural-science model that explains why a skill is or is not applied, not only whether it exists.
- Two questions each for Capability, Opportunity and Motivation, linked to the competencies where behavioural barriers matter most.
- Example: “My role gives me the opportunity to practice GenAI Tool Fluency regularly”. Responses use a 1 to 5 Likert scale.

- **Build assessment_responses** for the 55 focus-role employees x 16 questions x 2 cycles (1,760 rows). Proficiency answers are sampled around each employee’s required level, leaning slightly below it, so real gaps appear.
They show a small rise from prior_year to current_year.

- **In this phase:**
- paths use only the proficiency answers.
- COM-B answers are collected and held for later use.

**Note**

Row counts line up exactly (1,760 = 55 × 16 × 2), and the mean proficiency score shows a gentle upward drift from prior_year (2.47) to current_year (2.73).

This is a real, modest signal without artificial floor/ceiling effects. We will need this for Phase 3's **Predictive Readiness model**.

```python
import pandas as pd
import numpy as np
import random
import os

RAW_DIR = '/content/drive/MyDrive/Portfolio/LuminaLearningGroup/data/raw'

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

# ---------------------------------------------------------------
# C3.0. Load what we need from earlier clusters
# ---------------------------------------------------------------
employees = pd.read_csv(os.path.join(RAW_DIR, "employees.csv"))
role_competency_requirements = pd.read_csv(os.path.join(RAW_DIR, "role_competency_requirements.csv"))
competencies = pd.read_csv(os.path.join(RAW_DIR, "competencies.csv"))

FOCUS_ROLE_IDS = ["ROLE01", "ROLE04"]  # Instructional Design Lead, Solution Architect
focus_employees = employees[employees["role_id"].isin(FOCUS_ROLE_IDS)].copy()

# ---------------------------------------------------------------
# C3.1. assessment_questions (16 rows)
# ---------------------------------------------------------------
proficiency_questions = [
    {
        "question_id": f"Q{i+1:03d}",
        "question_text": f"Rate your current proficiency level in: {row.competency_name}",
        "question_type": "proficiency",
        "linked_competency_id": row.competency_id,
        "response_scale": "proficiency_scale (1-4)",
    }
    for i, row in enumerate(competencies.itertuples())
]

comb_links = {
    "comb_capability": ["COMP01", "COMP03"],
    "comb_opportunity": ["COMP07", "COMP08"],
    "comb_motivation": ["COMP01", "COMP04"],
}

comb_text_templates = {
    "comb_capability": "I have the knowledge/skill needed to apply {comp} in my daily work",
    "comb_opportunity": "My role/environment gives me the opportunity to practice {comp} regularly",
    "comb_motivation": "I am motivated to improve my {comp} skills",
}

comb_questions = []
q_counter = len(proficiency_questions) + 1
comp_name_lookup = dict(zip(competencies["competency_id"], competencies["competency_name"]))

for q_type, comp_ids in comb_links.items():
    for comp_id in comp_ids:
        comb_questions.append({
            "question_id": f"Q{q_counter:03d}",
            "question_text": comb_text_templates[q_type].format(comp=comp_name_lookup[comp_id]),
            "question_type": q_type,
            "linked_competency_id": comp_id,
            "response_scale": "1-5 Likert (free text, no dedicated lookup table)",
        })
        q_counter += 1

assessment_questions = pd.DataFrame(proficiency_questions + comb_questions)

# ---------------------------------------------------------------
# C3.2. assessment_responses
# ---------------------------------------------------------------
CYCLES = {
    "prior_year": "2025-09-01",
    "current_year": "2026-09-01",
}

rcr_lookup = {
    (row.role_id, row.competency_id): row.required_level
    for row in role_competency_requirements.itertuples()
}

def sample_prior_actual(required_level):
    val = np.random.normal(loc=required_level - 0.7, scale=0.9)
    return int(np.clip(round(val), 1, 4))

def sample_current_actual(prior_actual):
    improvement = 1 if random.random() < 0.35 else 0
    return int(np.clip(prior_actual + improvement, 1, 4))

response_rows = []
resp_counter = 1

proficiency_qs = assessment_questions[assessment_questions["question_type"] == "proficiency"]
comb_qs = assessment_questions[assessment_questions["question_type"] != "proficiency"]

for emp in focus_employees.itertuples():
    role_id = emp.role_id

    for q in proficiency_qs.itertuples():
        comp_id = q.linked_competency_id
        required_level = rcr_lookup.get((role_id, comp_id))
        if required_level is None:
            continue

        prior_val = sample_prior_actual(required_level)
        current_val = sample_current_actual(prior_val)

        for cycle, date, value in [("prior_year", CYCLES["prior_year"], prior_val),
                                     ("current_year", CYCLES["current_year"], current_val)]:
            response_rows.append({
                "response_id": f"RESP{resp_counter:06d}",
                "employee_id": emp.employee_id,
                "question_id": q.question_id,
                "response_value": value,
                "assessment_date": date,
                "assessment_cycle": cycle,
            })
            resp_counter += 1

    for q in comb_qs.itertuples():
        for cycle, date in CYCLES.items():
            response_rows.append({
                "response_id": f"RESP{resp_counter:06d}",
                "employee_id": emp.employee_id,
                "question_id": q.question_id,
                "response_value": random.randint(1, 5),
                "assessment_date": date,
                "assessment_cycle": cycle,
            })
            resp_counter += 1

assessment_responses = pd.DataFrame(response_rows)

# ---------------------------------------------------------------
# C3.3. Save
# ---------------------------------------------------------------
assessment_questions.to_csv(os.path.join(RAW_DIR, "assessment_questions.csv"), index=False)
assessment_responses.to_csv(os.path.join(RAW_DIR, "assessment_responses.csv"), index=False)

# ---------------------------------------------------------------
# C3.4. Sanity checks
# ---------------------------------------------------------------
print(f"assessment_questions: {len(assessment_questions)} rows")
print(assessment_questions["question_type"].value_counts())
print()
print(f"assessment_responses: {len(assessment_responses)} rows")
print()
print("Proficiency response_value distribution, prior vs current year:")
prof_q_ids = proficiency_qs["question_id"].tolist()
prof_resp = assessment_responses[assessment_responses["question_id"].isin(prof_q_ids)]
print(prof_resp.groupby("assessment_cycle")["response_value"].agg(["mean", "min", "max"]))
```

```python
import pandas as pd
import numpy as np
import random
import os

RAW_DIR = '/content/drive/MyDrive/Portfolio/LuminaLearningGroup/data/raw'
os.makedirs(RAW_DIR, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)

###############################
# WARNING (09 Oct 2026): do not re-run after NB02. It overwrites courses.csv and wipes NB02 columns (ksb_layer, duplicates).
###############################
# ---------------------------------------------------------------
# C4.0. Load reference tables generated earlier (Cluster 1)
# ---------------------------------------------------------------
ksbs = pd.read_csv(os.path.join(RAW_DIR, "ksbs.csv"))
competencies = pd.read_csv(os.path.join(RAW_DIR, "competencies.csv"))
comp_name_lookup = dict(zip(competencies["competency_id"], competencies["competency_name"]))

# ---------------------------------------------------------------
# C4.1. COURSE_SPEC — the generation parameters (ADR-0034). Nothing below
# this block is a literal course; every row is produced from these
# parameters, the same discipline used for employees.csv's ROLE_SPEC
# (Cluster 2). This replaces the old hand-written synthetic_courses
# list of 36 fixed rows.
# ---------------------------------------------------------------
PROVIDER_WEIGHTS = {
    # specialized (KSB-targeted) courses skew internal — LLG builds its own
    # AI/NLP/GenAI IP rather than buying it off the shelf (ADR-0031 logic)
    "specialized": {"Lumina_LMS": 0.65, "KonectIn": 0.15, "Ksera": 0.12, "Hiword": 0.08},
    # generic/orphan courses skew external — commodity soft-skill content
    # is licensed, not authored in-house
    "generic": {"KonectIn": 0.40, "Ksera": 0.35, "Hiword": 0.25},
}

COURSES_PER_KSB_RANGE = (3, 5)   # how many courses independently target each KSB
N_ORPHAN_COURSES = 20            # generic courses with no strong KSB match

TITLE_TEMPLATES = [
    "{topic}: {focus}",
    "Mastering {topic}",
    "Introduction to {topic}",
    "Advanced {topic} Techniques",
    "{topic} in Practice",
    "{topic} Fundamentals",
    "Applying {topic} to Real Projects",
    "{topic} for L&D Professionals",
]

FOCUS_WORDS_BY_TYPE = {
    "knowledge": ["Core Concepts", "Key Principles", "What You Need to Know"],
    "skill": ["Hands-On Practice", "Practical Application", "Building the Skill"],
    "behavior": ["Making It a Habit", "Putting It Into Practice", "Sustaining the Behavior"],
}

DESC_TEMPLATES = [
    "This course helps learners {action}, with practical examples drawn from real L&D projects.",
    "A focused module on how to {action}, built for working instructional designers and architects.",
    "Covers what it takes to {action}, through short lessons and applied exercises.",
    "Designed for learners who need to {action} as part of their day-to-day role.",
]

ORPHAN_TOPICS = [
    "Time Management for Busy Professionals",
    "Effective Meeting Facilitation Skills",
    "Workplace Communication Essentials",
    "Building Effective Presentations",
    "Cross-Cultural Collaboration Basics",
    "Introduction to Business Writing",
    "Conflict Resolution at Work",
    "Giving and Receiving Feedback",
    "Remote Team Collaboration",
    "Personal Productivity Systems",
]

FORMATS = ["video", "course", "job-aid", "checklist"]

ADVERB_PREFIXES = {"Iteratively", "Proactively"}


def _base_verb(word):
    """3rd-person-singular -> base form (understands -> understand,
    applies -> apply, manages -> manage). Handles slash-joined verbs
    like 'Flags/escalates' -> 'Flag/escalate'."""
    if "/" in word:
        return "/".join(_base_verb(w) for w in word.split("/"))
    if word.endswith("ies"):
        return word[:-3] + "y"
    if word.endswith("s") and not word.endswith("ss"):
        return word[:-1]
    return word


def ksb_to_action_phrase(ksb_text):
    """Convert a KSB statement (written 3rd-person, e.g. 'Understands X',
    'Can build Y', 'Iteratively reviews and refines Z') into an infinitive
    action phrase that reads correctly after 'helps learners' / 'need to'
    (e.g. 'understand X', 'build Y', 'iteratively review and refine Z')."""
    words = ksb_text.split(" ")

    if words[0] == "Can":
        rest = " ".join(words[1:])
        return rest[0].lower() + rest[1:] if rest else rest

    adverb = None
    idx = 0
    if words[0] in ADVERB_PREFIXES:
        adverb = words[0].lower()
        idx = 1

    words[idx] = _base_verb(words[idx])
    if idx + 2 < len(words) and words[idx + 1] == "and":
        words[idx + 2] = _base_verb(words[idx + 2])

    result = " ".join(words[idx:])
    if adverb:
        result = f"{adverb} {result}"
    return result[0].lower() + result[1:] if result else result


def sample_provider(kind):
    weights = PROVIDER_WEIGHTS[kind]
    providers, probs = zip(*weights.items())
    return random.choices(providers, weights=probs, k=1)[0]


def sample_duration(fmt):
    return {
        "video": random.randint(10, 30),
        "course": random.randint(45, 180),
        "job-aid": random.randint(5, 15),
        "checklist": random.randint(5, 10),
    }[fmt]


def make_title(topic, ksb_type, used_titles):
    template = random.choice(TITLE_TEMPLATES)
    focus = random.choice(FOCUS_WORDS_BY_TYPE[ksb_type])
    title = template.format(topic=topic, focus=focus)
    variant = 2
    base_title = title
    while title in used_titles:
        title = f"{base_title} (v{variant})"
        variant += 1
    used_titles.add(title)
    return title


# ---------------------------------------------------------------
# C4.2. Generate one course row per KSB target (3-5 independent courses each)
# ---------------------------------------------------------------
course_rows = []
counter = 1
target_ksb_for_row = []  # diagnostic only, not saved to courses.csv

for ksb in ksbs.itertuples():
    comp_name = comp_name_lookup[ksb.competency_id]
    action = ksb_to_action_phrase(ksb.ksb_text)
    n_courses = random.randint(*COURSES_PER_KSB_RANGE)
    used_titles = set()

    for _ in range(n_courses):
        title = make_title(comp_name, ksb.ksb_type, used_titles)
        desc = random.choice(DESC_TEMPLATES).format(action=action)
        provider = sample_provider("specialized")
        fmt = random.choice(FORMATS)

        course_rows.append({
            "course_id": f"CRS{counter:03d}",
            "course_title": title,
            "description_raw": desc,
            "course_source": provider,
            "is_real_excerpt": False,
            "credibility_sample_label": None,
            "duration_minutes": sample_duration(fmt),
            "format": fmt,
            "ksb_layer": None,
            "likely_duplicate_of": None,
            "core_learning_or_behavioral_enabler": None,
            "comb_blueprint": None,
            "target_ksb_id": ksb.ksb_id,                    # NEW (27 Sep 26)
            "target_competency_id": ksb.competency_id,      # NEW (27 Sep 26)
        })
        target_ksb_for_row.append(ksb.ksb_id)
        counter += 1

# ---------------------------------------------------------------
# C4.3. Orphan / generic courses — no strong KSB keyword overlap by design.
# Realistic that these are licensed off-the-shelf soft-skill content.
# ---------------------------------------------------------------
used_orphan_titles = set()
for _ in range(N_ORPHAN_COURSES):
    topic = random.choice(ORPHAN_TOPICS)
    title = topic
    variant = 2
    base_title = title
    while title in used_orphan_titles:
        title = f"{base_title} (v{variant})"
        variant += 1
    used_orphan_titles.add(title)

    fmt = random.choice(FORMATS)
    provider = sample_provider("generic")

    course_rows.append({
        "course_id": f"CRS{counter:03d}",
        "course_title": title,
        "description_raw": f"Practical, general-audience training on {topic.lower()}, applicable across roles and departments.",
        "course_source": provider,
        "is_real_excerpt": False,
        "credibility_sample_label": None,
        "duration_minutes": sample_duration(fmt),
        "format": fmt,
        "ksb_layer": None,
        "likely_duplicate_of": None,
        "core_learning_or_behavioral_enabler": None,
        "comb_blueprint": None,
        "target_ksb_id": None,              # NEW (27 Sep 26)
        "target_competency_id": None,       # NEW (27 Sep 26)
    })
    target_ksb_for_row.append(None)
    counter += 1

# ---------------------------------------------------------------
# C4.4. Real-excerpt credibility sample (ADR-0003) — placeholder rows.
# rows filled by hand with real abstracts:
#   -> arXiv:2508.19569 - Skill-based Explanations for Serendipitous Course Recommendation
#   -> arXiv:2505.02324 - From Course to Skill: Evaluating LLM Performance in Curricular Analytics
#   -> arXiv:2510.05495 - Automated Research Article Classification and Recommendation Using NLP and ML
#   -> arXiv:2007.11053 - Curriculum Vitae Recommendation
# ---------------------------------------------------------------
real_excerpt_placeholders = [
    "arXiv:2508.19569",
    "arXiv:2505.02324",
    "arXiv:2510.05495",
    "arXiv:2007.11053",
]
real_excerpt_ids = []
for arxiv_id in real_excerpt_placeholders:
    course_id = f"CRS{counter:03d}"
    real_excerpt_ids.append(course_id)
    course_rows.append({
        "course_id": course_id,
        "course_title": "[PASTE: real paper title]",
        "description_raw": f"[PASTE: abstract text from {arxiv_id} here — see ADR-0003]",
        "course_source": "arXiv",
        "is_real_excerpt": True,
        "credibility_sample_label": f"[PASTE: {arxiv_id} — title, abstract, license]",
        "duration_minutes": None,
        "format": "course",
        "ksb_layer": None,
        "likely_duplicate_of": None,
        "core_learning_or_behavioral_enabler": None,
        "comb_blueprint": None,
        "target_ksb_id": None,              # NEW (27 Sep 26)
        "target_competency_id": None,       # NEW (27 Sep 26)
    })
    target_ksb_for_row.append(None)
    counter += 1

courses = pd.DataFrame(course_rows)
# 09 Oct 2026
# Guard: courses.csv is enriched after generation (hand-pasted arXiv abstracts, NB02 ksb_layer
# and duplicate columns). Set REGENERATE_COURSES = True only to rebuild the catalog from scratch.
REGENERATE_COURSES = False
courses_path = os.path.join(RAW_DIR, "courses.csv")
if REGENERATE_COURSES or not os.path.exists(courses_path):
    courses.to_csv(courses_path, index=False)
    print("C4: courses.csv written to data/raw")
else:
    print("C4: courses.csv already exists and is protected (REGENERATE_COURSES = False). Not overwritten.")

# ---------------------------------------------------------------
# C4.5. Sanity checks
# ---------------------------------------------------------------
print(f"Total courses: {len(courses)}")
print(f"Synthetic: {(~courses['is_real_excerpt']).sum()}")
print(f"Real-excerpt placeholders: {courses['is_real_excerpt'].sum()} -> {real_excerpt_ids}")
print()
print("course_source distribution (synthetic only):")
print(courses[~courses["is_real_excerpt"]]["course_source"].value_counts())
print()
print("Format distribution:")
print(courses["format"].value_counts())
print()

target_ksb_series = pd.Series(target_ksb_for_row)
per_ksb_counts = target_ksb_series[target_ksb_series.notna()].value_counts()
print("Courses per targeted KSB (min/mean/max) — this is the redundancy signal NB02's")
print("duplicate detection should find, emerging from generation, not hand-placed:")
print(per_ksb_counts.agg(["min", "mean", "max"]))
print()
print(f"KSBs with coverage: {per_ksb_counts.shape[0]} of {len(ksbs)}")
print(f"Orphan (no target KSB) courses: {N_ORPHAN_COURSES}")
```
